## Qualidade dos dados

Nesta etapa, vamos verificar a presença de valores ausentes, registros duplicados e possíveis inconsistências nos datasets antes de iniciar o tratamento dos dados.

### Valores ausentes

```python
for nome, df in tabelas.items():
    print(f"\n{nome}")
    print(df.isnull().sum())
```

### Análise de valores ausentes

A análise de valores ausentes identificou problemas em três datasets.

No `df_pedidos`, foram encontrados valores ausentes nas colunas `order_approved_at` (160), `order_delivered_carrier_date` (1.783) e `order_delivered_customer_date` (2.965). Essas colunas representam diferentes etapas do processamento e da entrega dos pedidos, portanto os valores ausentes deverão ser investigados considerando o `order_status` antes de definir uma estratégia de tratamento.

No `df_avaliacoes`, foram identificados 87.656 valores ausentes em `review_comment_title` e 58.247 em `review_comment_message`. Como essas colunas representam comentários opcionais dos clientes, os valores ausentes não serão tratados automaticamente como erros.

No `df_produtos`, foram encontrados 610 valores ausentes nas informações de categoria e nas características textuais dos produtos, além de 2 valores ausentes nas dimensões e no peso dos produtos. Esses registros deverão ser avaliados durante a etapa de tratamento.

Os demais datasets (`df_clientes`, `df_itens`, `df_pagamentos` e `df_categorias`) não apresentam valores ausentes.

### Análise de registros duplicados

Nesta etapa, será verificada a existência de registros duplicados nos datasets, a fim de identificar possíveis repetições que possam afetar as análises posteriores.

```python
for nome, df in tabelas.items():
    print(f"{nome}: {df.duplicated().sum()} registros duplicados")
```

### Resultado da análise de registros duplicados

A análise não identificou registros completamente duplicados em nenhum dos datasets analisados. Portanto, não foram encontrados registros que precisem ser removidos por duplicidade nesta etapa.

É importante destacar que a ausência de linhas completamente duplicadas não significa que todos os identificadores sejam únicos em cada tabela. A repetição de identificadores pode ser esperada devido aos relacionamentos entre os datasets e será analisada posteriormente.

### Análise dos tipos de dados

Nesta etapa, serão avaliados os tipos de dados das colunas para identificar campos que possam exigir conversão ou padronização antes das análises.

```python
for nome, df in tabelas.items():
    print(f"\n{nome}")
    print(df.dtypes)
```

### Resultado da análise dos tipos de dados

A análise dos tipos de dados identificou que as principais colunas que exigem atenção são as relacionadas a datas.

No `df_pedidos`, as colunas `order_purchase_timestamp`, `order_approved_at`, `order_delivered_carrier_date`, `order_delivered_customer_date` e `order_estimated_delivery_date` estão armazenadas como `str`, embora representem datas e horários.

No `df_itens`, a coluna `shipping_limit_date` também está armazenada como `str`.

No `df_avaliacoes`, as colunas `review_creation_date` e `review_answer_timestamp` estão armazenadas como `str`.

Essas colunas deverão ser convertidas para um tipo de data apropriado durante a etapa de tratamento dos dados, permitindo a realização de análises temporais e cálculos relacionados aos prazos de entrega.

As principais colunas numéricas, como `price`, `freight_value`, `payment_value` e `review_score`, já estão armazenadas em tipos numéricos adequados. Os identificadores dos registros estão armazenados como `str`, o que é apropriado para esse tipo de informação.
